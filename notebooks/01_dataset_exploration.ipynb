```python
import os
import pandas as pd
import matplotlib.pyplot as plt
```

```python
DATA_PATH = "../data/raw/common_voice"

print("Dataset path:", DATA_PATH)
print("Exists:", os.path.exists(DATA_PATH))

if os.path.exists(DATA_PATH):
    print("\nFiles and folders:")
    for item in os.listdir(DATA_PATH):
        print(item)
```

```python
tsv_files = []

for root, dirs, files in os.walk(DATA_PATH):
    for file in files:
        if file.endswith(".tsv"):
            tsv_files.append(os.path.join(root, file))

print("TSV files found:")
for file in tsv_files:
    print(file)
```

```python
train_file = None

for file in tsv_files:
    if "train" in os.path.basename(file).lower():
        train_file = file
        break

print("Training file:", train_file)

if train_file:
    train_df = pd.read_csv(train_file, sep="\t")
    print("Number of samples:", len(train_df))
    print("\nColumns:")
    print(train_df.columns.tolist())
```

```python
clips_path = os.path.join(DATA_PATH, "clips")

print("Clips folder exists:", os.path.exists(clips_path))

if os.path.exists(clips_path):
    audio_files = os.listdir(clips_path)
    print("Number of audio files:", len(audio_files))
    print("First 10 audio files:")
    print(audio_files[:10])
```

```python
import librosa

if os.path.exists(clips_path) and len(audio_files) > 0:
    audio_file = os.path.join(clips_path, audio_files[0])

    audio, sample_rate = librosa.load(audio_file, sr=None)

    print("Audio file:", audio_file)
    print("Sample rate:", sample_rate)
    print("Duration:", len(audio) / sample_rate, "seconds")
```
